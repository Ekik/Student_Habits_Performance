```python
Tahap 1: Pembagian Data (Splitting)

Langkah 1 - Mengimpor library dan membaca dataset.
Library sklearn dan pandas diimpor, lalu file Student_Habits_Performance.csv dibaca ke dalam DataFrame df.

Mengubah data kategorikal menjadi angka.
Dataset disalin ke df_lencoder agar data asli tidak berubah. LabelEncoder kemudian mengubah semua kolom bertipe teks (object), seperti gender dan kualitas diet, menjadi angka, karena model machine learning hanya dapat memproses data numerik.

Memisahkan fitur dan target.
X (fitur): semua kolom kecuali social_media_hours.
y (target): kolom social_media_hours.
```

```python
import sklearn
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Membaca dataset
df = pd.read_csv('/content/Student_Habits_Performance.csv')

# Membuat df_lencoder dengan menyalin dataframe asli dan melakukan Label Encoding
df_lencoder = df.copy()
for col in df_lencoder.select_dtypes(include=['object']).columns:
    df_lencoder[col] = LabelEncoder().fit_transform(df_lencoder[col])

# Memisahkan fitur (X) dan target (y)
X = df_lencoder.drop(columns=['social_media_hours'])
y = df_lencoder['social_media_hours']
```

```python
Langkah 2 - Melihat sampel data.
Fungsi head() menampilkan lima baris pertama df_lencoder untuk memastikan hasil encoding sudah benar.
```

```python
# Menampilkan 5 baris pertama dari df_lencoder
df_lencoder.head()
```

```python
Langkah 3 - Membagi data latih dan data uji.
train_test_split membagi data dengan proporsi 80% data latih dan 20% data uji (test_size=0.2). Parameter random_state=1 membuat pembagian selalu sama setiap kali kode dijalankan, sehingga hasilnya dapat direproduksi.
```

```python
from sklearn.model_selection import train_test_split
# membagi dataset menjadi training dan testing
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
Langkah 4 - Memverifikasi jumlah data.
Jumlah data total, data latih, dan data uji dicetak untuk memastikan pembagiannya sesuai rasio 80:20.
```

```python
# menghitung panjang/jumlah data
print("Jumlah data: ",len(X))
# menghitung panjang/jumlah data pada x_train
print("Jumlah data latih: ",len(x_train))
# menghitung panjang/jumlah data pada x_test
print("Jumlah data test: ",len(x_test))
```

```python
Tahap 2: Pemodelan (Modelling)
Data latih digunakan untuk melatih tiga algoritma regresi yang berbeda, lalu kinerjanya dibandingkan.

Langkah 5 - Melatih tiga model.
Least Angle Regression (LARS) dengan n_nonzero_coefs=1. Model ini hanya memilih satu fitur yang paling berpengaruh, sehingga sangat sederhana.
Linear Regression, yaitu regresi linear standar yang menggunakan seluruh fitur dengan hubungan linear terhadap target.
Gradient Boosting Regressor (GBR), yaitu model ensemble yang membangun banyak pohon keputusan secara berurutan. Setiap pohon baru memperbaiki kesalahan pohon sebelumnya. Model ini mampu menangkap hubungan non-linear.
```

```python
#Melatih data yang sudah kita miliki dengan tiga algoritma yang berbeda
```

```python
# Melatih model 1 dengan algoritma Least Angle Regression
from sklearn import linear_model
lars = linear_model.Lars(n_nonzero_coefs=1).fit(x_train, y_train)
# Melatih model 2 dengan algoritma Linear Regression
from sklearn.linear_model import LinearRegression
LR = LinearRegression().fit(x_train, y_train)
# Melatih model 3 dengan algoritma Gradient Boosting Regressor
from sklearn.ensemble import GradientBoostingRegressor
GBR = GradientBoostingRegressor(random_state=184)
GBR.fit(x_train, y_train)
```

```python
Tahap 3: Evaluasi Model
Setiap model memprediksi data uji, lalu hasilnya dibandingkan dengan nilai sebenarnya menggunakan tiga metrik:

MAE (Mean Absolute Error)	Rata-rata selisih absolut antara prediksi dan nilai asli. Semakin kecil semakin baik.
MSE (Mean Squared Error)	Rata-rata kuadrat selisih. Lebih sensitif terhadap kesalahan besar. Semakin kecil semakin baik.
R² (R-squared)	Seberapa besar variasi target yang dapat dijelaskan model. Semakin mendekati 1 semakin baik.
```

```python
Langkah 6 - Evaluasi model LARS.
Prediksi dibuat pada data uji, ketiga metrik dihitung, lalu hasilnya disimpan dalam tabel df_results.
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
# Evaluasi pada model LARS
pred_lars = lars.predict(x_test)
mae_lars = mean_absolute_error(y_test, pred_lars)
mse_lars = mean_squared_error(y_test, pred_lars)
r2_lars = r2_score(y_test, pred_lars)
# Membuat dictionary untuk menyimpan hasil evaluasi
data = {
'MAE': [mae_lars],
'MSE': [mse_lars],
'R2': [r2_lars]
 }
# Konversi dictionary menjadi DataFrame
df_results = pd.DataFrame(data, index=['Lars'])
df_results
```

```python
Langkah 7 - Evaluasi model Linear Regression.
Prosesnya sama, dan hasilnya ditambahkan sebagai baris baru pada tabel df_results agar kedua model bisa dibandingkan langsung.
```

```python
# Evaluasi pada model Linear Regression
pred_LR = LR.predict(x_test)
mae_LR = mean_absolute_error(y_test, pred_LR)
mse_LR = mean_squared_error(y_test, pred_LR)
r2_LR = r2_score(y_test, pred_LR)
# Menambahkan hasil evaluasi LR ke DataFrame
df_results.loc['Linear Regression'] = [mae_LR, mse_LR, r2_LR]
df_results
```

```python
Langkah 8 - Membandingkan hasil.
Model dengan MAE dan MSE terkecil serta R² tertinggi dinilai paling akurat. Tabel df_results dipakai sebagai dasar memilih model terbaik.
```

```python
# Evaluasi pada model Linear Regression
pred_LR = LR.predict(x_test)
mae_LR = mean_absolute_error(y_test, pred_LR)
mse_LR = mean_squared_error(y_test, pred_LR)
r2_LR = r2_score(y_test, pred_LR)
# Menambahkan hasil evaluasi LR ke DataFrame
df_results.loc['Linear Regression'] = [mae_LR, mse_LR, r2_LR]
df_results
```

```python
Langkah 12 - Membuat API dengan Flask.
Aplikasi web sederhana dibangun dengan Flask:

Model dimuat saat aplikasi dijalankan.
Dibuat endpoint /predict dengan metode POST. Endpoint ini menerima data dalam format JSON, menjalankan prediksi dengan model, dan mengembalikan hasilnya dalam format JSON.
Fungsi find_free_port() mencari port yang kosong secara otomatis agar tidak terjadi bentrok port.
Server dijalankan di thread terpisah (background) sehingga tidak memblokir sel lain di Google Colab.
```

```python
from flask import Flask, request, jsonify
import joblib
import socket
import threading

# Inisialisasi aplikasi Flask
app = Flask(__name__)

# Memuat model yang telah disimpan dengan path lengkap
joblib_model = joblib.load('/content/gbr_model.joblib')

@app.route('/predict', methods=['POST'])
def predict():
    data = request.json['data']  # Mengambil data dari request JSON
    prediction = joblib_model.predict(data)  # Melakukan prediksi
    return jsonify({'prediction': prediction.tolist()})

def find_free_port():
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.bind(('', 0))
    port = s.getsockname()[1]
    s.close()
    return port

if __name__ == '__main__':
    free_port = find_free_port()
    print(f"Server Flask berjalan di background pada port: {free_port}")

    # Jalankan Flask Server di latar belakang agar tidak memblokir Colab
    flask_thread = threading.Thread(target=lambda: app.run(port=free_port, debug=False, use_reloader=False))
    flask_thread.daemon = True
    flask_thread.start()
```

```python
Langkah 13 – Menguji API.
Library requests mengirim data contoh ke endpoint /predict. Data berisi 15 fitur mahasiswa, di antaranya usia, jam belajar, jam menonton Netflix, persentase kehadiran, jam tidur, dan nilai ujian. Program menampilkan status code dan hasil prediksi. Status 200 berarti API berjalan dengan baik dan model berhasil memberikan prediksi jam media sosial untuk mahasiswa tersebut.
```

```python
import requests

# Menyesuaikan port dengan server Flask yang sedang aktif
url = f'http://127.0.0.1:{free_port}/predict'

# Payload pengujian dengan 15 fitur Student Habits
payload = {
    "data": [[
        0,    # student_id
        23,   # age
        0,    # gender
        0.0,  # study_hours_per_day
        1.1,  # netflix_hours
        0,    # part_time_job
        85.0, # attendance_percentage
        8.0,  # sleep_hours
        0,    # diet_quality
        6,    # exercise_frequency
        2,    # parental_education_level
        0,    # internet_quality
        8,    # mental_health_rating
        1,    # extracurricular_participation
        56.2  # exam_score
    ]]
}

try:
    response = requests.post(url, json=payload)
    print('Status Code:', response.status_code)
    print('Hasil Prediksi:', response.json())
except Exception as e:
    print('Gagal terhubung ke server Flask:', e)
```

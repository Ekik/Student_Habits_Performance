Mean Squared Error (MSE)
yang juga disebut Mean Squared Deviation, adalah selisih kuadrat antara nilai aktual dan nilai prediksi. Dengan kata lain, MSE menunjukkan seberapa dekat garis terbaik yang sesuai dengan kumpulan titik. MSE selalu bernilai positif. Kuadrat diambil untuk menghilangkan tanda negatif. Semakin dekat MSE ke 0 semakin akurat prediksinya.

Mean Absolute Error (MAE), yang juga Kesalahan absolut rata-rata, memberi kita rata-rata selisih absolut antara nilai yang diamati dan nilai yang diprediksi. Perbedaan antara MAE dan MSE adalah bahwa MAE mengambil selisih absolut antara nilai yang diprediksi dan nilai yang diamati, sedangkan MSE mengambil selisih kuadratnya.

R2 (Koefisien Determinasi) digunakan untuk mengukur seberapa baik model regresi menjelaskan variasi data. Nilai terbaiknya adalah 1,0, yang berarti prediksi model sempurna. Nilai 0,0 menunjukkan model tidak lebih baik daripada memprediksi rata-rata data aktual, sedangkan nilai negatif berarti model lebih buruk. Jika nilai aktual konstan, skor R2 dapat menjadi NaN atau -Inf. Secara default, Scikit-learn menggantinya menjadi 1,0 jika prediksi sempurna dan 0,0 jika prediksi tidak sempurna.
